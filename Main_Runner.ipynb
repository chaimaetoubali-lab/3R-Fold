```python
import os
import sys
import pandas as pd

if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

os.chdir(PROJECT_ROOT)

from data.build_dataset import build_dataset
from models.pretrained_embedder import PretrainedRNAEmbedder
from evaluation.benchmark import run_benchmark, split_dataset

query_ids = set(pd.read_csv(f"{PROJECT_ROOT}/results/ablation_query_ids.csv")["query_id"])

dataset = list(build_dataset())
kb_data, test_data = split_dataset(dataset, test_fraction=0.2, seed=42)

# keep exact same subset
test_data = [x for x in test_data if x["id"] in query_ids]

embedder = PretrainedRNAEmbedder(device="cpu", max_length=256)

methods = [
    dict(method_name="vienna_only", retrieval_type="none", use_guidance=False, guidance_mode="consensus", k=0),
    dict(method_name="rnafm_faiss_no_guidance_k5", retrieval_type="faiss", use_guidance=False, guidance_mode="consensus", k=5),
    dict(method_name="rnafm_faiss_guided_k1", retrieval_type="faiss", use_guidance=True, guidance_mode="consensus", k=1),
    dict(method_name="rnafm_faiss_guided_k3", retrieval_type="faiss", use_guidance=True, guidance_mode="consensus", k=3),
    dict(method_name="rnafm_faiss_guided_k5", retrieval_type="faiss", use_guidance=True, guidance_mode="consensus", k=5),
    dict(method_name="rnafm_faiss_guided_k7", retrieval_type="faiss", use_guidance=True, guidance_mode="consensus", k=7),
    dict(method_name="rnafm_faiss_guided_k14", retrieval_type="faiss", use_guidance=True, guidance_mode="consensus", k=14),
    dict(method_name="rnafm_faiss_template_guided_k5", retrieval_type="faiss", use_guidance=True, guidance_mode="template_transfer", k=5),
    dict(method_name="oracle_guided_k5", retrieval_type="oracle", use_guidance=True, guidance_mode="consensus", k=5),
]

all_rows = []

for cfg in methods:
    print("Running:", cfg["method_name"])
    df = run_benchmark(
        dataset=dataset,
        embedder=embedder,
        n_queries=None,
        test_fraction=0.2,
        seed=42,
        method_name=cfg["method_name"],
        retrieval_type=cfg["retrieval_type"],
        use_guidance=cfg["use_guidance"],
        guidance_mode=cfg["guidance_mode"],
        k=cfg["k"],
        batch_size=4,
        max_length=256,
        cache_path="results/kb_embeddings_rnafm_len256.pt",
    )

    # keep exact query subset only
    df = df[df["query_id"].isin(query_ids)].copy()
    all_rows.append(df)

ablation_df = pd.concat(all_rows, ignore_index=True)
ablation_df.to_csv(f"{PROJECT_ROOT}/results/ablation_results_fixed50.csv", index=False)

summary = (
    ablation_df.groupby("method", as_index=False)
    .agg(
        n=("f1", "count"),
        f1=("f1", "mean"),
        sens=("sensitivity", "mean"),
        ppv=("ppv", "mean"),
    )
)

summary.to_csv(f"{PROJECT_ROOT}/results/ablation_summary_fixed50.csv", index=False)

print(summary.round(4))
```

```python
import pandas as pd
import numpy as np
import os
from config import get_results_path

ROOT = get_results_path()

bench = pd.read_csv(f"{ROOT}/benchmark_main.csv")
ab = pd.read_csv(f"{ROOT}/ablation_results.csv")
dsr = pd.read_csv(f"{ROOT}/dsr_global_all_results.csv")
hyb = pd.read_csv(f"{ROOT}/dsr_faiss_hybrid_results.csv")

# Optional LOFO if available
lofo_path = f"{ROOT}/lofo_summary.csv"
lofo = pd.read_csv(lofo_path) if os.path.exists(lofo_path) else None

vienna = bench[bench["method"] == "vienna_only"].copy()
faiss = ab[ab["method"] == "rnafm_faiss_guided_k14"].copy()
dsr14 = dsr[dsr["method"] == "rnafm_dsr_global_guided_k14"].copy()
hyb14 = hyb[hyb["method"] == "rnafm_dsr_faiss_guided_k14"].copy()

final_df = pd.concat([vienna, faiss, dsr14, hyb14], ignore_index=True)


method_map = {
    "vienna_only": "ViennaRNA",
    "rnafm_faiss_guided_k14": "RNA-FM + FAISS",
    "rnafm_dsr_global_guided_k14": "RNA-FM + DSR",
    "rnafm_dsr_faiss_guided_k14": "3R-Fold : RNA-FM + DSR + FAISS guided",
}
order = ["ViennaRNA", "RNA-FM + FAISS", "RNA-FM + DSR", "3R-Fold"]

final_df["method_clean"] = final_df["method"].map(method_map)

# Save the combined dataframe used for paper tables
final_df.to_csv(f"{ROOT}/final_family_benchmark_methods.csv", index=False)

print("Combined methods:")
print(final_df["method_clean"].value_counts())
print()


overall = (
    final_df.groupby("method_clean", as_index=False)
    .agg(
        n=("f1", "count"),
        f1_mean=("f1", "mean"),
        f1_std=("f1", "std"),
        sensitivity_mean=("sensitivity", "mean"),
        ppv_mean=("ppv", "mean"),
    )
)

overall["method_clean"] = pd.Categorical(overall["method_clean"], categories=order, ordered=True)
overall = overall.sort_values("method_clean").reset_index(drop=True)

print("=== OVERALL BENCHMARK TABLE ===")
print(overall.round(4))
print()
```

```python

overall.to_csv(f"{ROOT}/paper_overall_benchmark_table.csv", index=False)


family_table = (
    final_df.groupby(["family", "method_clean"], as_index=False)
    .agg(
        n=("f1", "count"),
        f1_mean=("f1", "mean"),
        sensitivity_mean=("sensitivity", "mean"),
        ppv_mean=("ppv", "mean"),
    )
)

family_wide = family_table.pivot(index="family", columns="method_clean", values="f1_mean").reset_index()

# keep desired order of methods if present
family_cols = ["family"] + [m for m in order if m in family_wide.columns]
family_wide = family_wide[family_cols]

# Optional gain columns over ViennaRNA
if "ViennaRNA" in family_wide.columns:
    for m in ["RNA-FM + FAISS", "RNA-FM + DSR", "3R-Fold"]:
        if m in family_wide.columns:
            family_wide[f"Gain_{m}"] = family_wide[m] - family_wide["ViennaRNA"]

print("=== FAMILY BENCHMARK TABLE ===")
print(family_wide.round(4))
print()

family_wide.to_csv(f"{ROOT}/paper_family_benchmark_table.csv", index=False)

# Create length bins from the final_df rows
bins = [0, 100, 300, 700, 100000]
labels = ["<100", "100-299", "300-699", "700+"]
final_df["length_bin"] = pd.cut(final_df["length"], bins=bins, labels=labels)

length_summary = (
    final_df.groupby(["method_clean", "length_bin"], observed=False, as_index=False)
    .agg(
        n=("f1", "count"),
        f1_mean=("f1", "mean"),
        f1_std=("f1", "std"),
        sensitivity_mean=("sensitivity", "mean"),
        ppv_mean=("ppv", "mean"),
    )
)

length_wide = length_summary.pivot(index="length_bin", columns="method_clean", values="f1_mean").reset_index()

length_order = ["<100", "100-299", "300-699", "700+"]
length_wide["length_bin"] = pd.Categorical(length_wide["length_bin"], categories=length_order, ordered=True)
length_wide = length_wide.sort_values("length_bin")

length_cols = ["length_bin"] + [m for m in order if m in length_wide.columns]
length_wide = length_wide[length_cols]

# Optional gain columns
if "ViennaRNA" in length_wide.columns:
    for m in ["RNA-FM + FAISS", "RNA-FM + DSR", "3R-Fold"]:
        if m in length_wide.columns:
            length_wide[f"Gain_{m}"] = length_wide[m] - length_wide["ViennaRNA"]

print("=== LENGTH BENCHMARK TABLE ===")
print(length_wide.round(4))
print()

length_wide.to_csv(f"{ROOT}/paper_length_benchmark_table.csv", index=False)
length_summary.to_csv(f"{ROOT}/length_summary.csv", index=False)


if lofo is not None:
    print("=== LOFO TABLE ===")
    print(lofo.round(4))
    print()
    lofo.to_csv(f"{ROOT}/paper_lofo_table.csv", index=False)

print("Saved files:")
print(f"{ROOT}/final_family_benchmark_methods.csv")
print(f"{ROOT}/paper_overall_benchmark_table.csv")
print(f"{ROOT}/paper_family_benchmark_table.csv")
print(f"{ROOT}/paper_length_benchmark_table.csv")
if lofo is not None:
    print(f"{ROOT}/paper_lofo_table.csv")
```
